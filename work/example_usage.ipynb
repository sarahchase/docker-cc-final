The function get_data_from_csv will read your .csv file and give you a list of dictionaries. You can specify which columns you are interested in using the variable columns_list. Replace this with your file name and column names. The function will print the first row of your data so you can see that it's working as you expect.

```python
data = get_data_from_csv(file_name = "spotify_top_1000_tracks.csv", 
                         columns_list = ["track_name", 
                                         "artist",
                                         "release_date", 
                                         "duration_min"])
```

Your data is now stored in the variable called data. However, all values are stores as STRINGS so we might need to convert some of them to numbers. The function convert_dates will handle any columns where the value is a date. 

When working with dates in Python, we have to specify how the date is formatted. We do this with something called a format string. For dates in the form YYYY-MM-DD we use "%Y-%m-%d". You can read about possible format string options here:
https://www.w3schools.com/python/gloss_python_date_format_codes.asp

This also works with time (minutes, hours, seconds). In the next function, we can send a list of format strings if the data is not consistent. In the Spotify data, for example, some rows only have the year. So I said in my list that it should first try the format string "%Y-%m-%d" and then, if that doesn't work, try "%Y". If neither works, it will throw an error. You can have a longer list of possibilities if you want. 

Call the function with your data and the name of the column you want to modify. The function returns your dataset with new columns based on how it parsed the date. If you have multiple columns that need to be read, run the function multiple times.

```python
data = convert_dates(data, "release_date", ["%Y-%m-%d", "%Y"])
```

Next, we will convert numerical data. The first reader function shows the data as a string. I have to specify which column should be converted to floats and call the convert_numbers function. If you need to do this with multiple columns, call the function multiple times.

```python
data = convert_numbers(data, "duration_min")
```

Let's look at the first row of our data to see that everything is the right format. Notice that new columns (keys) were created to store the numerical data. The variable data is a list. Every item in the list is a dictionary.

This next part will be specific to you! Here's an example with both qualitative and quantitative data.

Let's say I want to count how many times each artist appears on the list. I know some artists have more than 1 song on the list. I'm going to make a dictionary to store how many times each artist appears. Then I'm going to use a for loop to go through the rows in the data and check the artist for each one. This next block does not use any functions from the module. This part you should code yourself using what you know about lists, dictionaries, loops, and conditional statements.

```python
# Initialize an empty dictionary
artist_counts = {}

# Look at each row one at a time
for row in data:
    # Get the name of the artist for that row
    name = row["artist"]
    # If the artist has shown up before, add one to their count
    if name in artist_counts:
        artist_counts[name] += 1
    # If this is the first time, create a new key for them and start with 1
    else:
        artist_counts[name] = 1
```

Next, I want to see which artists appear the most. Because this is a little complicated, I made a function for you. The function sort_by_value will take a dictionary of artist names and counts and sort by count. The function returns a list of tuples where each tuple is the name and count of the artist. The list is sorted by the count of songs that artist has in the data.

```python
# Call the function to sort the dictionary data
artists_sorted = sort_by_value(artist_counts)

# Show the 5 top artists
n = 5
print(f"The {n} artists with most songs on the list are\n")

# This loop only goes up to 5
for i in range(n):
    # Print data for each artist one at a time
    print(artists_sorted[i])
```

In order to make this interactive, I want the user to choose how many artists I show on the plot. Above I chose 5 just to practice printing. Here's a function to make a bar plot. The function takes one input n: the number of artists to show.

```python
def make_bar_plot_artists(n):
    
    names = [item[0] for item in artists_sorted[:n]]
    counts = [item[1] for item in artists_sorted[:n]]

    plt.figure()
    plt.bar(names, counts, color="green", edgecolor="black")
    plt.xticks(rotation=90) # Rotate the artist names 90 degrees
    plt.xlabel("Artists")
    plt.ylabel("Number of Songs")
    plt.title("Top 1000 Most-Played Spotify Songs of All Time")
    plt.show()
```

Now I need to set up an interactive widget and link that widget to my function. Look online for different widget and plotting options.

```python
# When the user chooses a value on the slider, n_slider equals that value
n_slider = widgets.IntSlider(
    value=5, # Default value
    min=1,
    max=20,
    step=1, # Step size
    description='Number of artists to show:',
    continuous_update=False,
    style={'description_width': 'auto'}, # Don't cut off label
    layout=widgets.Layout(width='500px') # Make it wide
)

# Connect the widget to the function. Specify that the input variable n should go with the value of n_slider
output = widgets.interactive_output(make_bar_plot_artists, {'n': n_slider})

# Display the slider and the plot
display(n_slider, output)
```

Now let's do an example with numerical data. There are built-in functions for making histograms given a list of numbers. Pull out all the song durations from our data using list comprehension.

```python
durations = [row["duration_min_float"] for row in data]
```

Now we can send this list into the histogram function. The number of bins determines how detailed the plot is. Move the slider to see what happens.

```python
def plot_duration_hist(n_bins):
    plt.hist(durations, bins=n_bins, edgecolor="white")
    plt.xlabel("Duration (minutes)")
    plt.ylabel("Number of Songs")
    plt.show()
```

This time the plotting function is linked to a different slider. Note that the input name n_bins is provided

```python
# Whatever value the user chooses is assigned to the variable bins_slider
# Starts with a default of 10
bins_slider = widgets.IntSlider(
    value=10,
    min=2,
    max=40,
    step=1,
    description="Number of histogram bins:",
    continuous_update=False,
    style={"description_width": "auto"},
    layout=widgets.Layout(width="500px")
)

# Name of function: plot_duration_hist
# Name of function input: n_bins
# Variable with function input: bins_slider
output = widgets.interactive_output(plot_duration_hist, {"n_bins": bins_slider})

display(bins_slider, output)
```
